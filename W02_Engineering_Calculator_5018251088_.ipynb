```python
### 1. Speed Unit Conversion

speed_kmh = float(input("enter speed [km/h] : "))
speed_ms = speed_kmh / 3.6
print("")
print(f"speed {speed_ms:.2f} m/s")
```

```python
### 2. Power Unit Conversion

power_kw = float(input("Power [kW] : "))
power_w = power_kw * 1000
print("")
print(f"Power = {power_w:.2f} W")
```

```python
### 3. Electrical Power and Energy Conversion

voltage_v = float(input("Voltage [v] : "))
current_A = float(input("Current [A] : "))
operating_time_h = float(input("Operating Time (hour) : "))

power_w = voltage_v * current_A
energy_wh = power_w * operating_time_h

print("")
print("==== ELECTRICAL CALCULATION ====")
print(f"Voltage = {voltage_v:.2f} V")
print(f"Current = {current_A:.2f} A")
print(f"Power = {power_w:.2f} W")
print(f"Operating Time = {operating_time_h:.2f} hour")
print(f"Energy = {energy_wh:.2f} Wh")
```

```python
### 4. Comparison and Logical Operators
#### Test Case 1

voltage_v = float(input("Voltage [v] : "))
current_A = float(input("Current [A] : "))
operating_time_h = float(input("Operating Time (hour) : "))

power_w = voltage_v * current_A
energy_wh = power_w * operating_time_h

power_limit = 200
voltage_limit = 24

is_high_power = power_w > power_limit
is_high_voltage = voltage_v > voltage_limit

print("")
print("==== ELECTRICAL CALCULATION ====")
print(f"Voltage = {voltage_v:.2f} V")
print(f"Current = {current_A:.2f} A")
print(f"Power = {power_w:.2f} W")
print(f"Operating Time = {operating_time_h:.2f} hour")
print(f"Energy = {energy_wh:.2f} Wh")

print("")
print(f"High Power : {is_high_power}")
print(f"High Voltage : {is_high_voltage}")
print(f"Both Conditions : {is_high_power and is_high_voltage}")
print(f"Any Conditions : {is_high_power or is_high_voltage}")
```

```python
### Short Reflection

1. Apa perbedaan "int" dan "float"?
    Tipe data integer untuk bilangan bulat tanpa koma/desimal, sedangkan tipe data floating point untuk bilangan desimal yang memiliki tanda koma/titik desimal.

2. Mengapa hasil "input()" perlu dikonversi menggunakan "float()" untuk perhitungan numerik?
    Di Python selalu mengembalikan data dalam bentuk teks/string, meskipun pengguna memasukkan angka. String tidak bisa langsung digunakan untuk operasi matematika perhitungan. Konversi menggunakan floating point mengubah teks angka tersebut menjadi tipe numerik desimal agar bisa dihitung secara matematis.

3. Apa perbedaan operator = dan == dalam phyton?
    = "Assignment": Digunakan untuk memasukkan atau menyimpan nilai ke dalam suatu variabel, sedangkan == "Equality Comparison": Digunakan untuk membandingkan apakah dua nilai bernilai sama.

4. Apa fungsi operator "and" dan "or"?
    "and" Menghasilkan "True" hanya jika semua kondisi yang dibandingkan bernilai "True". "or": Menghasilkan "True" jika salah satu atau seluruh kondisi yang dibandingkan bernilai "True".

5. Apa satu kesalahan yang Anda temui saat mengerjakan tugas ini, dan bagaimana Anda memperbaikinya?
    typo pada code sehingga munculnya error, dapat diselesaikan dengan membaca log error untuk mencari kesalahan atau typo
```
